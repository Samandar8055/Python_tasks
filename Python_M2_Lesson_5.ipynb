## Homework 2:

```python
import pandas as pd

data = {'Month': ['Jan','Feb','Mar','Apr'],
        'Sales': [5000,6000,7500,8000],
        'Expenses': [3000,3500,4000,4500]}

sales_and_expenses = pd.DataFrame(data)
sales_and_expenses
```

```python
#Calculate and display the maximum sales and expenses.
sales_and_expenses[['Sales','Expenses']].max()

```

```python
# Calculate and display the minimum sales and expenses.
sales_and_expenses[['Sales','Expenses']].min()
```

```python
# Calculate and display the average sales and expenses.
sales_and_expenses[['Sales','Expenses']].mean()
```

## Homework 3

```python
# Homework 3
# task 1

import pandas as pd

data = {'Category': ['Rent','Utilities','Groceries','Entertainment'],
        'January': [1200,200,300,150],
        'February': [1300,220,320,160],
        'March':[1400,240,330,170],
        'April':[1500,250,350,180]}

h3_df = pd.DataFrame(data)
h3_df
```

```python
# task 2
# Calculate and display the maximum expense for each category.

h3_df['Max_expenses']=h3_df[['January', 'February', 'March', 'April']].max(axis=1)
h3_df[['Category','Max_expenses']]
```

```python
# task 3
# Calculate and display the minimum expense for each category.

h3_df['Min_Expenses'] = h3_df[['January', 'February', 'March', 'April']].min(axis=1)
h3_df[['Category','Min_Expenses']]
```

```python
# task 4
# Calculate and display the average expense for each category.

h3_df['Avrage_Expenses'] = h3_df[['January', 'February', 'March', 'April']].mean(axis=1)
h3_df[['Category','Avrage_Expenses']]
h3_df
```
